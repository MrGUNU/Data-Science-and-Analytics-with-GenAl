```python
d = {}
print(type(d))
```

```python
d = {1,2,3}
print(type(d))
```

```python
d = {1:10, 2:20}
print(type(d))
```

```python
# For access the value we have to call the key. (key can be any kind of data type)
d = {10:100, 20:200, 30:300}
print(d[10])
```

```python
# It has mutable in nature.
d = {10:100, 20:200, 30:300}
d[10] = 1000

print(d)
```

```python
# 1st way to create a dictionary
d = dict(name = "Bhoumik", age = 21, gender = "male")
print(d)
```

```python
# 2nd way to create dictionary
d = dict([("name", "Bhoumik"), ("age", 21)])
print(d)
```

```python
# Dictionary Traversing 1st way
a = {10:100, 20:200, 30:300, 40:400}

for i in a:
    print(a[i])
```

```python
# 2nd way
a = {10:100, 20:200, 30:300, 40:400}

for i in a.values():
    print(i)
```
