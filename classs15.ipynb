```python
# MAP function

a = [1,2,3,4]

l = map(lambda x : x**2, a)
print(list(l))
```

```python
def square(x):
    return x**2\

a = [1,2,3,4]
l = map(square, a)
print(list(l))
```

```python
# Filter

a = [1,2,3,4,5,6]

l = []

for i in a:
    if i%2 == 0:
        l.append(i)
print(l)
```

```python
# Filter using lambda expression

a = [1,2,3,4,5,6]

l = filter(lambda x: x%2 == 0, a)
print(list(l))
```
