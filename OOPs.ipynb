```python
# Class

class Factory:
    a = 12   # attribute

    def hello():   # method
        print("How are you")  

print(Factory.a)
Factory.hello()
```

```python
# Objects

class Factory:
    a = "Hello I am an attribute"

    def hello():
        print("Hello I am a method")

obj = Factory()  # Obj becomes an object who can access anything inside the class till now

print(obj.a)
```

```python
# Constructor

class Factory():
    def __init__(self,material,zips,pockets):
        self.material = material
        self.zips = zips
        self.pockets = pockets

obj = Factory("Leather",3,3)

print(obj.zips)
```

```python
# In depth Attributes and Methods

class Animal:
    gender = "Male"  # class attribute

    def __init__(self,name,age):
        self.name = name  # instance attribute
        self.age = age    # instance attribute

    def info(self):     # instance method
        print("This is a method")

    @classmethod
    def clmethod(cls):      # class method
        print(f"{cls.gender} is your gender")

    @staticmethod
    def hello():   # static method
        print("Hello I am a static method")


obj = Animal("Lion", 12)
obj.info()
obj.clmethod()
obj.hello()
```

```python
# Make a student registration system ask for name, age, number, blood group  register 3 students

class Regestration:
    def __init__(self,name,age,number,blood):
        self.name = name
        self.age = age
        self.number = number
        self.blood = blood

    def info(self):
        print(f"Your name is {self.name} \nYour age is {self.age}\nYour number is {self.number}\nYour blood group is {self.blood}")


s1 = Regestration("Tony", 47, 893264, "B+")
s2 = Regestration("Thor", 50, 938457, "A-")
s3 = Regestration("Steve", 49, 243875, "AB+")

s1.info()
s2.info()
s3.info()
```

```python
# Inheritance

class Animal:           #Parent class / Super class
    def __init__(self,name,age):
        self.name = name
        self.age = age

    def info(self):
        print(f"Your name is {self.name} and age is {self.age}")

class Human(Animal):    #Child class / Sub class
    pass

obj1 = Animal("Lion", 12)
obj2 = Human("Tony", 21)

obj1.info()
obj2.info()
```

```python
# Encapsulation

class Animal:
    __name = "lion"

    def speak(self):
        print("hello I will roar")


class Human(Animal):
    def say(self):
        print(f"hello my name is {super().__name} ")

obj = Human()

obj.say()
```

```python
# Dunder methods

class Students:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def __str__(self):
        return f"{self.name} is your name and your marks is {self.marks}"

obj = Students("Gunu", 95)

print(obj)
```

```python
class Shopping:
    def __init__(self, items):
        self.items = items

    def __len__(self):
        return len(self.items)

obj = Shopping(['apple', 'orange', 'banana', 'milk'])
obj2 = Shopping(['apple', 'milk'])

print(len(obj))
print(len(obj2))
```
